```python
# libraries

from google.colab import drive
from ultralytics import YOLO
import cv2
from google.colab.patches import cv2_imshow
import os
import pandas as pd
from IPython.display import display, HTML
import random
from sklearn.metrics import classification_report, accuracy_score
from sklearn.metrics import precision_score, recall_score, f1_score
```

Training the model

```python
# training

model = YOLO("yolov8n-cls.pt")
data_path = "/content/drive/MyDrive/combined_GI_gastrohun"
model.train(data=data_path, epochs=30, imgsz=224, batch=32, name= 'yolov8n-GI')
```

<br>
<font  size=5>
Testing on validation set

```python
# validation 

model = YOLO("/content/runs/classify/yolov8n-GI/weights/best.pt")
val_path = "/content/drive/MyDrive/GI_data_split/val"
class_names = ["abnormal", "normal"]

results_list = []
sample_images = []


for class_name in class_names:
    class_path = os.path.join(val_path, class_name)
    for img_file in os.listdir(class_path):
        if img_file.lower().endswith(('.jpg', '.jpeg', '.png')):
            img_path = os.path.join(class_path, img_file)

            results = model(img_path)
            pred_class = class_names[results[0].probs.top1]
            confidence = results[0].probs.top1conf.item()
            is_correct = (pred_class == class_name)


            results_list.append({
                'image': img_file,
                'true_class': class_name,
                'pred_class': pred_class,
                'confidence': confidence,
                'correct': is_correct
            })


            if random.random() < 0.1 and len(sample_images) < 10:
                img = cv2.imread(img_path)
                img = cv2.resize(img, (300, 300))
                sample_images.append({
                    'image': img,
                    'filename': img_file,
                    'true': class_name,
                    'pred': pred_class,
                    'conf': confidence
                })


results_df = pd.DataFrame(results_list)

print(f"\nEvaluation Summary for {len(results_df)} images:")
print(f"Accuracy: {(results_df['correct'].mean()*100):.2f}%")

print("\nSample Predictions:")
for sample in sample_images:
    display(HTML(f"""
    <div style='margin:20px; border:1px solid #ddd; padding:10px; border-radius:5px;'>
        <h4>{sample['filename']}</h4>
        <p>True: <b>{sample['true']}</b> | Predicted: <b>{sample['pred']}</b> (Confidence: {sample['conf']:.2f})</p>
    </div>
    """))
    cv2_imshow(sample['image'])
    print("-"*50)


print("\nFirst 10 Results:")
display(results_df.head(10))
```

store the results for better analysis and future usage

<br>
<font  size=5>
Generating results table

```python
# metrices 

y_true = results_df['true_class']
y_pred = results_df['pred_class']


print("\nClassification Report:")
print(classification_report(y_true, y_pred, target_names=class_names))

precision = precision_score(y_true, y_pred, average='binary', pos_label='abnormal')
recall = recall_score(y_true, y_pred, average='binary', pos_label='abnormal')
f1 = f1_score(y_true, y_pred, average='binary', pos_label='abnormal')
accuracy = accuracy_score(y_true, y_pred)

print(f"\nMetrics (class 'abnormal' as positive):")
print(f"Accuracy:  {accuracy:.4f}")
print(f"Precision: {precision:.4f}")
print(f"Recall:    {recall:.4f}")
print(f"F1 Score:  {f1:.4f}")
```

Testing on abnormal set

```python
# testing on abnormal set

model = YOLO("/content/drive/MyDrive/best (5).pt")

folder_path = "/content/drive/MyDrive/testset_GI_Data/abnormal"
class_names = ["abnormal", "normal"]  


image_files = [f for f in os.listdir(folder_path) if f.lower().endswith(('.jpg', '.jpeg', '.png'))]

for file_name in image_files:
    image_path = os.path.join(folder_path, file_name)

    
    results = model(image_path)
    pred_index = results[0].probs.top1
    pred_class = class_names[pred_index]
    confidence = results[0].probs.top1conf.item()

    
    if pred_class == "normal":
        print(f"{file_name} → {pred_class} ({confidence:.2f})")

        img = cv2.imread(image_path)
        img_resized = cv2.resize(img, (300, 300))
        cv2.putText(img_resized, f'{pred_class} ({confidence:.2f})', (10, 25),
                    cv2.FONT_HERSHEY_SIMPLEX, 0.8, (0, 255, 0), 2)
        cv2_imshow(img_resized)
```

Testing on normal set

```python
# testing on normal set

model = YOLO("/content/drive/MyDrive/best (5).pt")

image_dir = "/content/drive/MyDrive/normal"  
class_names = ["abnormal", "normal"]


image_extensions = ['.jpg', '.jpeg', '.png', '.bmp']
image_paths = [os.path.join(image_dir, f) for f in os.listdir(image_dir)
              if os.path.splitext(f)[1].lower() in image_extensions]

for image_path in image_paths:

        
        results = model(image_path)
        pred_index = results[0].probs.top1
        pred_class = class_names[pred_index]
        confidence = results[0].probs.top1conf.item()

        # Print results
        print(f"\nImage: {os.path.basename(image_path)}")
        print(f"Predicted class: {pred_class}")
        print(f"Confidence: {confidence:.2f}")
```
